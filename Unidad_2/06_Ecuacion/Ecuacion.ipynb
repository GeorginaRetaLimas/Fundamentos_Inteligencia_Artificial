```python
import numpy as np
import matplotlib.pyplot as plt

# Configuración
PREC = 0.001
x = np.arange(0, 1 + PREC, PREC)

# Función
f_x = np.sin(5 * np.pi * x)**6 * np.exp(-2 * np.log(2) * ((x - 0.1) / 0.8))

# Tabla de valores
print(f"{'x':<10} {'f(x)':<15}")
print("-" * 25)

for xi, yi in zip(x, f_x):
    print(f"{xi:<10.3f} {yi:<15.8f}")

# Gráfica
plt.figure(figsize=(10, 6))
plt.plot(x, f_x, color='blue', label='f(x)')

plt.title('Gráfica de f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.xlim(0, 1)
plt.grid(True)
plt.legend()

plt.show()
```
