```python
# Negación (NOT)

def neg(p: bool) -> bool:
    return False if p else True

print("Negación:", neg(True))
# Falso

print("Negación:", neg(False))
# Verdadero


# Conjunción (AND)

def y(p: bool, q: bool) -> bool:
    return p and q


# Disyunción inclusiva (OR)

def o(p: bool, q: bool) -> bool:
    return p or q


# Disyunción exclusiva (XOR)

def ox(p: bool, q: bool) -> bool:
    return neg(q) if p else q

print("Disyunción exclusiva:", ox(True, True))
# Falso

print("Disyunción exclusiva:", ox(True, False))
# Verdadero

print("Disyunción exclusiva:", ox(False, True))
# Verdadero

print("Disyunción exclusiva:", ox(False, False))
# Falso


# Implicación Lógica (Condicional)

def imp(p: bool, q: bool) -> bool:
    return False if (p and not q) else True

print("Implicación:", imp(True, False))
# Falso

print("Implicación:", imp(True, True))
# Verdadero

print("Implicación:", imp(False, False))
# Verdadero


# Doble Implicación (Bicondicional)

def ssi(p: bool, q: bool) -> bool:
    return p == q

print("Doble implicación:", ssi(True, True))
# Verdadero

print("Doble implicación:", ssi(True, False))
# Falso

print("Doble implicación:", ssi(False, True))
# Falso

print("Doble implicación:", ssi(False, False))
# Verdadero


# ===============================================
# El prompt del usuario promedio
# "Da acceso si el admin o la contraseña es correcta"

print("🚑 Auditoria de ingenieria de prompts 🚑\n")

llm_interpretacion_ambigua = lambda p, q, r: o(p, y(q, neg(r)))

# CASO B:

llm_interpretacion_estructurada = lambda p, q, r: y(o(p, q), neg(r))

entorno_ataque = (True, False, True)

acceso_caso_a = llm_interpretacion_ambigua(*entorno_ataque)

acceso_caso_b = llm_interpretacion_estructurada(*entorno_ataque)

print("SIMULACIÓN DE ATAQUE DE SEGURIDAD")

print("Escenario: Admin = True | Contraseña Correcta = False | Cuenta suspendida = True\n")

print("Resultado con prompt ambiguo : {' 🔒 ACCESO CONCEDIDO (Fallo de seguridad)' if acceso_caso_a else' 🔒 ACCESO DENEGADO'} ")

print("Resultado con Prompt Estricto: {' 🔒 ACCESO CONCEDIDO (Fallo de seguridad)' if acceso_caso_b else' 🔒 ACCESO DENEGADO (Sistema seguro)'} ")

```
